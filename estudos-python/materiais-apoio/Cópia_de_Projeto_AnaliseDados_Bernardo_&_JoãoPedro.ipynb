**Informações sobre o dataset de câncer de pulmão:**

Este conjunto de dados contém informações relacionadas à mortalidade por câncer de pulmão, reunindo de forma abrangente dados clínicos, demográficos e históricos de pacientes diagnosticados com a doença. O objetivo principal é permitir a análise de fatores que podem influenciar a progressão do câncer, a resposta ao tratamento e a taxa de sobrevivência dos pacientes.

| Coluna              | Descrição |
|---------------------|----------------------------------------------------------------------------------------|
| id                  | Identificador único de cada paciente no conjunto de dados                             |
| age                 | Idade do paciente no momento do diagnóstico                                            |
| gender              | Gênero do paciente (ex.: masculino, feminino)                                         |
| country             | País ou região onde o paciente reside                                                 |
| diagnosis_date      | Data em que o paciente foi diagnosticado com câncer de pulmão                         |
| cancer_stage        | Estágio do câncer de pulmão no momento do diagnóstico (ex.: Estágio I, II, III, IV)   |
| family_history      | Indica se há histórico familiar de câncer (ex.: sim, não)                             |
| smoking_status      | Status de tabagismo do paciente (ex.: fumante atual, ex-fumante, nunca fumou, passivo)|
| bmi                 | Índice de Massa Corporal (IMC) do paciente no momento do diagnóstico                   |
| cholesterol_level   | Nível de colesterol do paciente                                                        |
| hypertension        | Indica se o paciente tem hipertensão (pressão alta) (ex.: sim, não)                   |
| asthma              | Indica se o paciente tem asma (ex.: sim, não)                                         |
| cirrhosis           | Indica se o paciente tem cirrose hepática (ex.: sim, não)                             |
| other_cancer        | Indica se o paciente já teve outro tipo de câncer além do diagnóstico principal       |
| treatment_type      | Tipo de tratamento que o paciente recebeu (ex.: cirurgia, quimioterapia, radioterapia, combinado) |
| end_treatment_date  | Data em que o paciente concluiu o tratamento ou faleceu                               |
| survived            | Indica se o paciente sobreviveu (ex.: sim, não)

# **Tratamento dos dados**

**Verificação de valores nulos:**

```python
print(df.isnull().sum())
#não há dados nulos
```

**Verificação de duplicatas:**

```python
print(df.duplicated().sum())
#não há dados duplicados
```

**Conversão de variáveis categóricas:**

```python
# Padronizando colunas de texto (remove espaços e coloca primeira letra maiúscula)
df['smoking_status'] = df['smoking_status'].str.strip().str.title()
df['treatment_type'] = df['treatment_type'].str.strip().str.title()
df['family_history'] = df['family_history'].str.strip().str.title()
df['cancer_stage'] = df['cancer_stage'].str.strip().str.title()

# Agora sim o mapeamento funciona:
df['smoking_status'] = df['smoking_status'].map({
    'Never Smoked': 0,
    'Passive Smoker': 1,
    'Former Smoker': 2,
    'Current Smoker': 3
})

df['treatment_type'] = df['treatment_type'].map({
    'Surgery': 0,
    'Chemotherapy': 1,
    'Radiation Therapy': 2,
    'Combined': 3
})

df['family_history'] = df['family_history'].map({'Yes': 1, 'No': 0})

df['cancer_stage'] = df['cancer_stage'].map({
    'Stage I': 1,
    'Stage Ii': 2,
    'Stage Iii': 3,
    'Stage Iv': 4
})
```

#Plotagens

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style="whitegrid")
```


Sobrevivência


```python
sns.countplot(x='survived', data=df, hue='survived', palette='rocket', legend=False)
plt.title("Distribuição de Sobreviventes")
plt.xticks([0,1], ['Não', 'Sim'])
plt.show()
```

Distribuição de Idades ( Box Plot )

```python
plt.figure(figsize=(8, 8))
sns.boxplot(data=df, y='age',)
plt.title('Distribuição de Idades (Box Plot)')
plt.ylabel('Idade')
plt.show()
```

Distribuição de Idades ( Histograma )

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='age', kde=True, bins=30,hue='age', palette='rocket', legend=False)
plt.title('Distribuição de Idade dos Pacientes')
plt.xlabel('Idade')
plt.ylabel('Frequência')
plt.show()
```

Distribuição de BMI

```python
sns.histplot(df['bmi'], bins=30, kde=True,)
plt.title('Distribuição de BMI')
plt.xlabel('BMI')
plt.ylabel('Frequência')
plt.show()
```

Sobrevivência por Estágio do Câncer

```python
sns.countplot(x='cancer_stage', hue='survived', data=df, palette= 'rocket')
plt.title('Sobrevivência por Estágio do Câncer')
plt.xlabel('Estágio do Câncer')
plt.ylabel('Contagem')
plt.legend(title='Sobreviveu', labels=['Não', 'Sim'])
plt.show()
```

# Medidas descritivas


Média de Idade e BMI por Sobrevivência

Taxa de Sobrevivência por Gênero

Taxa de Sobrevivência por Status de Tabagismo

```python
df_smoking = pd.read_csv('Lung Cancer.csv')
df_smoking['survived'] = df['survived']
df_smoking.groupby('smoking_status')['survived'].mean()
```



Conclusões iniciais

* A idade e o BMI parecem ter uma leve influência na sobrevivência, com sobreviventes tendo uma média de idade e BMI ligeiramente menores.

* O estágio do câncer é um fator crucial, com taxas de sobrevivência diminuindo drasticamente em estágios mais avançados.

* O gênero não parece ser um fator determinante na sobrevivência para este conjunto de dados.

* O status de tabagismo mostra uma correlação com a sobrevivência, com fumantes ativos tendo a menor taxa de sobrevivência.
